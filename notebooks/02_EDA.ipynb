```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(
    "../dataset/cleaned_supply_chain_data.csv",
    encoding="latin1"
)

print("Cleaned dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Summary Statistics")
df.describe()
```

```python
print("Mean:")
print(df.mean(numeric_only=True))
```

```python
print("Median:")
print(df.median(numeric_only=True))
```

```python
print("Standard Deviation:")
print(df.std(numeric_only=True))
```

```python
numeric_columns = df.select_dtypes(include="number").columns.tolist()

print("Numerical columns:")
print(numeric_columns)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Product_Price"], bins=20)

plt.xlabel("Product Price")
plt.ylabel("Number of Orders")
plt.title("Distribution of Product Prices")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Days_for_shipment_scheduled"],
    df["Days_for_shipping_real"],
    alpha=0.3
)

plt.xlabel("Scheduled Shipment Days")
plt.ylabel("Actual Shipping Days")
plt.title("Scheduled vs Actual Shipping Days")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df["Benefit_per_order"].dropna())

plt.ylabel("Benefit per Order")
plt.title("Benefit per Order - Outlier Analysis")

plt.show()
```

```python
correlation_columns = [
    "Days_for_shipping_real",
    "Days_for_shipment_scheduled",
    "Benefit_per_order",
    "Sales_per_customer",
    "Late_delivery_risk",
    "Order_Item_Quantity",
    "Product_Price",
    "Sales",
    "Order_Profit_Per_Order"
]

correlation_matrix = df[correlation_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Supply Chain Feature Correlation Matrix")

plt.show()
```

```python
df["Shipping_Delay_Days"] = (
    df["Days_for_shipping_real"]
    - df["Days_for_shipment_scheduled"]
)

df["Shipping_Delay_Days"].describe()
```

```python
print(
    "Delayed shipments:",
    (df["Shipping_Delay_Days"] > 0).sum()
)

print(
    "On-time/early shipments:",
    (df["Shipping_Delay_Days"] <= 0).sum()
)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Shipping_Delay_Days"],
    bins=20
)

plt.xlabel("Shipping Delay (Days)")
plt.ylabel("Number of Shipments")
plt.title("Distribution of Shipping Delay")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Shipping_Delay_Days"],
    bins=20
)

plt.xlabel("Shipping Delay (Days)")
plt.ylabel("Number of Shipments")
plt.title("Distribution of Shipping Delay")

plt.show()
```

## EDA Findings

1. The dataset contains shipment, order, customer, product, sales, and delivery-related information.

2. Mean, median, and standard deviation were calculated for the numerical variables to understand their distribution.

3. The scheduled shipment duration was compared with the actual shipping duration to understand shipment delays.

4. Product price distribution was analyzed using a histogram.

5. A scatter plot was used to study the relationship between scheduled and actual shipping duration.

6. Boxplot analysis was used to identify potential outliers in the data.

7. A correlation matrix was used to understand relationships between important numerical features.

8. Shipping delay was calculated as:
   
   Actual Shipping Days - Scheduled Shipment Days

9. These findings provide a basis for the machine-learning model used for shipment delay-risk prediction.
